# Problema #1: Calculadora de propina inteligente

## Problema

Crear un programa que solicite el total de una cuenta y el porcentaje de propina que desea dejar el cliente. El programa debe calcular el monto de la propina y el total final a pagar.

## Entrada

- Total de la cuenta
porcentaje de propina.

## Proceso

- Convertir los datos ingresados a tipo float.

- Calcular la propina:

propina = total_cuenta * (porcentaje / 100)

- Calcular total final:

total_final = total_cuenta + propina

## Salida esperada

- Total de la cuenta.
- Monto de la propina.
- Total final a pagar.

```python

try:
    total_cuenta = float(input("Ingrese el total de la cuenta: "))
    porcentaje = float(input("Ingrese el porcentaje de propina: "))

    propina = total_cuenta * (porcentaje / 100)
    total_final = total_cuenta + propina

    print(f"\nTotal de la cuenta: ₡{total_cuenta:.2f}")
    print(f"Propina: ₡{propina:.2f}")
    print(f"Total a pagar: ₡{total_final:.2f}")

except ValueError:
    print("Error: Debe ingresar valores numéricos.")
```

# Problema #2: Detector de número par/impar

## Problema

Crear un programa que solicite un número entero y determine si es par o impar.

## Entrada

- Un número entero.

## Proceso

- Convertir el dato a tipo int.
- Utilizar el operador módulo %.
- Si el residuo es 0, el número es par.
De lo contrario, es impar.

## Salida esperada

- Mostrar un mensaje indicando si el número es par o impar.

```python
try:
    numero = int(input("Ingrese un número entero: "))

    if numero % 2 == 0:
        print(f"El número {numero} es par.")
    else:
        print(f"El número {numero} es impar.")

except ValueError:
    print("Error: Debe ingresar un número entero válido.")
```

# Problema #3: Clasificador de edad

## Problema

Crear un programa que solicite la edad y clasifique a la persona según su rango etario.

## Entrada

- Edad de la persona.

## Proceso

- Convertir la entrada a tipo int.
- Evaluar la edad:
Menor de 18 años → Menor de edad.
Entre 18 y 64 años → Adulto.
65 años o más → Adulto mayor.

## Salida esperada

- Mostrar la categoría correspondiente a la edad ingresada.

```python
try:
    edad = int(input("Ingrese su edad: "))

    if edad < 18:
        categoria = "Menor de edad"
    elif edad < 65:
        categoria = "Adulto"
    else:
        categoria = "Adulto mayor"

    print(f"Clasificación: {categoria}")

except ValueError:
    print("Error: Debe ingresar una edad válida.")

```

# Problema #4: Validador de compra con descuento

## Problema

Crear un programa que solicite el nombre de un cliente, el monto de su compra y si es cliente frecuente. Dependiendo de estas condiciones, se aplicará un descuento.v

## Entrada

- Nombre del cliente.
- Monto de compra.
- Indicación de si es cliente frecuente (sí/no).

## Proceso

- Solicitar los datos.
Convertir el monto a float.
- Si el cliente es frecuente y la compra es mayor o igual a ₡50 000, aplicar un 10% de descuento.
- En caso contrario, no aplicar descuento.
- Calcular el monto final.

## Salida esperada

- Nombre del cliente.
- Monto original.
- Descuento aplicado.
- Total a pagar.

## Pruebas manuales

Ejercicio 1

Entrada

Cuenta: 20000
Propina: 10

Salida esperada

Propina: ₡2000
Total a pagar: ₡22000
Ejercicio 2

Entrada

Número: 15

Salida esperada

El número 15 es impar.
Ejercicio 3

Entrada

Edad: 70

Salida esperada

Clasificación: Adulto mayor.
Ejercicio 4

Entrada

Cliente: Ana
Compra: 60000
Cliente frecuente: sí

Salida esperada

Descuento: ₡6000
Total a pagar: ₡54000

```python
try:
    cliente = input("Ingrese el nombre del cliente: ").strip().title()
    monto = float(input("Ingrese el monto de la compra: ₡"))
    frecuente = input("¿Es cliente frecuente? (si/no): ").strip().lower()

    descuento = 0

    if frecuente == "si" and monto >= 50000:
        descuento = monto * 0.10

    total_pagar = monto - descuento

    print("\n--- Resumen de Compra ---")
    print(f"Cliente: {cliente}")
    print(f"Monto original: ₡{monto:.2f}")
    print(f"Descuento aplicado: ₡{descuento:.2f}")
    print(f"Total a pagar: ₡{total_pagar:.2f}")

except ValueError:
    print("Error: Debe ingresar un monto válido.")
```
